```python
#lngest subarray with sumk (ps)#2pointer/sliding window applicable for +ve
arr=list(map(int,input().split()))
k=int(input('enter a number'))
n=len(arr)
left,right,max_len=0,0,0
sum=arr[0]
while right<n:
    while left<right and sum>k:
        sum-=arr[left]
        left+=1
    if sum==k:
        max_len=max(max_len,right-left+1)
    right+=1
    if right<n:
        sum+=arr[right]
print(max_len)
#o(n)sc o(1)

```

```python
#longest subarray sums k=0 prefix sum o(n)and o(1)
arr=list(map(int,input().split()))
k=int(input('enter a number'))
n=len(arr)
prefix_sum=0
first_seen={}
max_len=0
for i in range(n):
    prefix_sum+=arr[i]
    if prefix_sum==0:
        max_len=i+1
    if prefix_sum in first_seen:
        length= i-first_seen[prefix_sum]
        max_len=max(max_len,length)
    if prefix_sum not in first_seen:
        first_seen[prefix_sum]=i
print(max_len)
```

```python
#no of subarrays with xor =k
arr=list(map(int,input().split()))
k=int(input('enter a number'))
freq={0:1}
prefix_xor=0
cnt=0
for i in arr:
    prefix_xor^=i
    targeet=prefix_xor^k
    if targeet in freq:
        cnt+=freq[targeet]
    freq[prefix_xor]=freq.get(prefix_xor,0)+1
print(cnt)
```

```python
#find repeating and missing number
arr=list(map(int,input().split()))
n=len(arr)
s=0
s2=0
sn=n*(n+1)//2
s2n=n*(n+1)*(2*n+1)//6
for i in arr:
    s+=i
    s2+=i*i
v1=s-sn
v2=s2-s2n
v2=v2//v1
x=(v1+v2)//2
y=x-v1
print(f'repeating no is {x} and missing no is {y}')

```

```python
#same qs with xor approach
arr=list(map(int,input().split()))
n=len(arr)
xor=0
for i in range(n):
    xor^=arr[i]
    xor^=i+1
number=(xor & ~(xor-1) )
zero=0
one=0
for i in arr:
    if (i & number)!=0:
        one^=i
    else:
        zero^=i
for i in range(1,n+1):
    if (i & number)!=0:
        one^=i
    else:
        zero^=i
cnt=0
for i in arr:
    if i==zero:
        cnt+=1
if cnt==2:
    print(f'repeating no is {zero} and missing no is {one}')
else:
    print(f'repeating no is {one} and missing no is {zero}')
```

```python
#count inversion with merge sort intuition
def merge(arr,low,mid,high):
    temp=[]
    left=low
    right=mid+1
    cnt=0
    while left<=mid and right<=high:
        if arr[left]<=arr[right]:
            temp.append(arr[left])
            left+=1
        else:
            temp.append(arr[right])
            cnt+=(mid-left+1)
            right+=1
    while left<=mid:
        temp.append(arr[left])
        left+=1
    while right<=high:
        temp.append(arr[right])
        right+=1
    for i in range(low,high+1):
        arr[i]=temp[i-low]
    return cnt
def merge_sort(arr,low,high):
    cnt=0
    if low<high:
        mid=(low+high)//2
        cnt+=merge_sort(arr,low,mid)
        cnt+=merge_sort(arr,mid+1,high)
        cnt+=merge(arr,low,mid,high)
    return cnt
def count_inversions(arr):
    return merge_sort(arr,0,len(arr)-1)
```
